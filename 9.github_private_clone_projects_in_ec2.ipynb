To **clone a private GitHub repository on an EC2 instance** using a **GitHub personal access token (PAT)**, follow these steps:

---

### ✅ 1. **Create a GitHub Personal Access Token (PAT)**

1. Go to [https://github.com/settings/tokens](https://github.com/settings/tokens)
2. Click **"Generate new token (classic)"** (or "Fine-grained token" if preferred)
3. Select scopes like:
   - `repo` (for full control of private repositories)
4. Set an expiration (e.g., 30 days)
5. Click **Generate token**
6. **Copy the token immediately** – you won’t be able to see it again.

---

### ✅ 2. **SSH into your EC2 instance**

```bash
ssh -i your-key.pem ec2-user@your-ec2-public-ip
```

---

### ✅ 3. **Install Git (if not installed)**

```bash
sudo yum update -y      # For Amazon Linux
sudo yum install git -y
```

---

### ✅ 4. **Clone the private repo using HTTPS and your token**

Format:

```bash
git clone https://<your-username>:<your-token>@github.com/<your-username-or-org>/<repo-name>.git
```

**Example:**

```bash
git clone https://your-username:ghp_ABC123yourtokenhere@github.com/your-username/my-private-repo.git
```

> 🔐 This exposes the token in plain text. Use this **only temporarily**, or consider the SSH key method for long-term use (see below).

---

### 🔐 Optional: Clone Private Repo Securely (using SSH instead)

If you'd prefer a more secure long-term setup:

#### 1. Generate SSH Key on EC2
```bash
ssh-keygen -t rsa -b 4096 -C "your-email@example.com"
# Press Enter through all prompts
```

#### 2. Copy the public key
```bash
cat ~/.ssh/id_rsa.pub
```

#### 3. Add the public key to your GitHub:
- Go to **GitHub → Settings → SSH and GPG keys**
- Click **New SSH Key**
- Paste the key, give it a name, and save.

#### 4. Test SSH connection
```bash
ssh -T git@github.com
```

#### 5. Clone repo using SSH
```bash
git clone git@github.com:your-username/your-private-repo.git
```
