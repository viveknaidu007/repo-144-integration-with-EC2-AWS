```python
pm2 start index.js -i max --name adtip-server
(max auto-detects CPU cores (so 2 in your case).)

-i max	Best for your setup (auto 2 workers)
-i 2	Safe manual config
-i 4+	Not useful on 2-core machine (can degrade performance)
```

```python
Use 1 worker per CPU core for maximum efficiency.

You have 2 vCPUs, so:

    pm2 start index.js -i 2 --name adtip-server

 What if you give -i 8 on a 2-core machine?

    PM2 will create 8 workers anyway, but:

        They’ll compete for CPU, causing context switching.

        It can actually slow down performance under load.

        Only useful if your app is I/O-bound and super lightweight
```

```python
The error you're getting is due to a missing system dependency (bzip2) on your EC2 Ubuntu instance. This is required for PhantomJS to extract .tar.bz2 files during the install process.
```

# solution

```python
#Install bzip2 and try again
sudo apt-get update && sudo apt-get install -y bzip2

#Then try installing your Node dependencies again:
npm install



#info:

Why this happens only on EC2 and not your local machine?

Because most Ubuntu EC2 instances come minimal — they don't have common utilities like bzip2, zip, unzip, etc., pre-installed. Your local machine probably already has bzip2.
```
