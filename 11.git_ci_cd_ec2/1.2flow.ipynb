Example GitHub Actions Workflow (Optional)

If you’re setting up a CI/CD pipeline to deploy to your EC2 instance, here’s a basic example of how these secrets might be used in a GitHub Actions workflow:

```python
name: Deploy to EC2
on:
  push:
    branches:
      - main
jobs:
  deploy:
    runs-on: ubuntu-latest
    steps:
      - name: Checkout code
        uses: actions/checkout@v3
      - name: Deploy to EC2
        env:
          EC2_HOST: ${{ secrets.EC2_HOST }}
          EC2_USER: ${{ secrets.EC2_USER }}
          EC2_KEY: ${{ secrets.EC2_KEY }}
        run: |
          echo "$EC2_KEY" > private_key.pem
          chmod 600 private_key.pem
          ssh -o StrictHostKeyChecking=no -i private_key.pem ${EC2_USER}@${EC2_HOST} << 'EOF'
            # Commands to run on EC2 instance
            cd ~/ci-cd-backend
            git pull origin main
            # Add your deployment commands (e.g., restart server)
          EOF
          rm private_key.pem
```
