```python
Check your .env file:
text

DB_HOST=localhost  # Or an IP like 3.6.15.198
DB_USER=root
DB_PASSWORD=yourpassword
DB_NAME=yourdb
```

```python
#Check if MySQL is Installed on EC2:

mysql --version
```

```python
#If not installed, install it:

sudo apt update
sudo apt install mysql-server
sudo systemctl start mysql
sudo systemctl enable mysql
```

```python
#Set Up MySQL Database:

sudo mysql -u root -p
```

```python
#Create a database and user:

CREATE DATABASE adtip_db;
CREATE USER 'adtip_user'@'localhost' IDENTIFIED BY 'yourpassword';
GRANT ALL PRIVILEGES ON adtip_db.* TO 'adtip_user'@'localhost';
FLUSH PRIVILEGES;
EXIT;
```

```python
#errros: if we got :

Error: ER_NOT_SUPPORTED_AUTH_MODE: Client does not support authentication protocol requested by server; consider upgrading MySQL client

sudo mysql -u root -p

#modify 
ALTER USER 'adtip_user'@'localhost' IDENTIFIED WITH 'mysql_native_password' BY 'yourpassword';
FLUSH PRIVILEGES;
EXIT;


Explanation:

    mysql_native_password: Switches the authentication plugin to one compatible with your Node.js mysql client.
    'yourpassword': Must match DB_PASSWORD in your .env (currently "yourpassword").
    FLUSH PRIVILEGES: Applies the changes immediately.
```

```python
# creating users table :

sudo mysql -u root -p

USE adtip_db;
CREATE TABLE users (
  id INT AUTO_INCREMENT PRIMARY KEY,
  mobile_number VARCHAR(15) NOT NULL UNIQUE,
  name VARCHAR(255),
  email VARCHAR(255) UNIQUE,
  password VARCHAR(255),
  created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
EXIT;
```

```python
#Ensure your .env matches:

DB_HOST=localhost
DB_USER=adtip_user
DB_PASSWORD=yourpassword
DB_NAME=adtip_db
```

```python
#Verify MySQL is Running:
sudo systemctl status mysql

#If not running, start it:
sudo systemctl start mysql
```

```python
If Using Remote MySQL (e.g., RDS):

    Update DB_HOST to the RDS endpoint (e.g., mydb.rds.amazonaws.com).
    Ensure the EC2 security group allows outbound traffic to MySQL port 3306.
    Ensure the RDS security group allows inbound traffic from 3.6.15.198 on port 3306.
```
