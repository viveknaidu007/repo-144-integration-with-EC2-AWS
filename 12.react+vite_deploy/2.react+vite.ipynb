```python
Step 3: Install a Web Server (Nginx) on EC2

sudo yum update -y  # For Amazon Linux
sudo apt update -y  # For Ubuntu

sudo yum install nginx -y  # For Amazon Linux
sudo apt install nginx -y  # For Ubuntu

sudo systemctl start nginx
sudo systemctl enable nginx

sudo systemctl status nginx
```

```python
Configure Nginx to Serve Your React App

    #Edit the Nginx configuration file:
sudo nano /etc/nginx/nginx.conf  # For Amazon Linux
sudo nano /etc/nginx/sites-available/default  # For Ubuntu


#Replace or update the server block to serve the dist folder and handle React's client-side
server {
    listen 80;
    server_name <your-ec2-public-ip> <your-domain>;  # Replace with your IP or domain

    root /usr/share/nginx/html/dist;
    index index.html;

    location / {
        try_files $uri $uri/ /index.html;  # Handle React routing
    }

    error_page 404 /index.html;
}

//
Replace <your-ec2-public-ip> with your EC2 public IP.
If you have a domain, include it in server_name
```

```python
Save and exit (Ctrl+O, Enter, Ctrl+X in nano).
```

```python
#Troubleshooting

    App not loading: Check Nginx logs:
    bash

sudo tail -f /var/log/nginx/error.log
Permission issues: Ensure the dist folder has correct ownership (nginx or www-data).
Port not accessible: Verify your EC2 security group allows inbound traffic on port 80 (HTTP) and 443 (HTTPS).
React routing issues: Ensure the try_files directive in the Nginx config is correctly set to redirect to index.html.
```

```python
Summary of Key Commands:

# Local: Build React + Vite
npm run build

# Local: Transfer files to EC2
scp -i <your-key.pem> -r ./dist ec2-user@<ec2-public-ip>:/home/ec2-user/

# EC2: Install Nginx
sudo yum install nginx -y  # or sudo apt install nginx -y
sudo systemctl start nginx
sudo systemctl enable nginx

# EC2: Move files
sudo mv /home/ec2-user/dist /usr/share/nginx/html/
sudo chown -R nginx:nginx /usr/share/nginx/html/dist
sudo chmod -R 755 /usr/share/nginx/html/dist

# EC2: Configure Nginx
sudo nano /etc/nginx/nginx.conf  # or /etc/nginx/sites-available/default
sudo nginx -t
sudo systemctl restart nginx
```
