# To run a React + Vite frontend application on an EC2 instance

```python
cd /path/to/your/repo
npm install
```

```python
#Run the Development Server with nohup: Use nohup to run the npm run dev command in the background, so it persists after you exit the SSH session
nohup npm run dev &


/
nohup: Prevents the process from terminating when the terminal session ends.
&: Runs the command in the background.
Output will be redirected to a file named nohup.out in the current directory by default.
```

```python
#Verify the Process is Running:
ps aux | grep vite

#Or, check the nohup.out file for logs:
cat nohup.out
```

```python
Access the Application:

    By default, Vite’s development server listens on localhost.
     To access it externally (e.g., from your browser), you need to make it listen on 0.0.0.0:
        Edit your vite.config.js or vite.config.ts in the project root to include:


vite.config.js or .ts:
export default {
  server: {
    host: '0.0.0.0',
    port: 5173, // or your preferred port
  },
}
```

```python
Ensure your EC2 instance’s Security Group allows inbound traffic on the port Vite is using (e.g., 5173):

    Go to the AWS EC2 Console > Security Groups > Edit Inbound Rules.
    Add a rule for Custom TCP, port 5173, source 0.0.0.0/0 (or restrict to your IP for security).
```

```python
Redirect Logs (Optional): If you want to redirect logs
 to a specific file instead of nohup.out, modify the nohup command:


nohup npm run dev > myapp.log 2>&1 &

/
> myapp.log: Redirects standard output to myapp.log.
2>&1: Redirects standard error to the same file.
```

```python
Manage the Running Process

#Check the Process ID (PID):
ps aux | grep vite

#Note the PID of the process running npm run dev.
Stop the Process (if needed):

kill <pid>

#Restart the Server (after config changes)
kill <pid>
nohup npm run dev &
```

# production

```python
Considerations for Production:

    Development Server: npm run dev is meant for development and not optimized for production. For production, build the app and serve it with a proper server:
    bash

npm run build

#Serve the built files (from the dist folder) using a web server like nginx or serve:

npm install -g serve
nohup serve -s dist -l 5173 &

#Port: Ensure the port (e.g., 5173) is open in the EC2 Security Group and not blocked by a firewall.
#Process Management: For long-term use, consider using a process manager like pm2 instead of nohup:


npm install -g pm2
pm2 start npm --name "vite-app" -- run dev
pm2 save
pm2 startup

This ensures the app restarts on system reboot and provides better process monitoring
```

# Errors:

```python
Troubleshooting

    Port Not Accessible: Verify the Security Group settings and ensure Vite is configured to listen on 0.0.0.0.
    Process Stops: Check nohup.out or myapp.log for errors (e.g., missing dependencies or configuration issues).
    Node Version Issues: Ensure the Node.js version on EC2 matches your project’s requirements. Use nvm to manage Node versions:
    bash

    curl -o- https://raw.githubusercontent.com/nvm-sh/nvm/v0.39.5/install.sh | bash
    nvm install <version>
    nvm use <version>

If you encounter specific errors or need help with production setup
```
