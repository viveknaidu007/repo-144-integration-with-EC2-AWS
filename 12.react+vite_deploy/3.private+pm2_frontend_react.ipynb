# To run a React + Vite frontend application on an EC2 instance in the background using nohup (so it continues running even after you disconnect from the SSH session),

```python
cd /path/to/your/repo
npm install
nohup npm run dev &
ps aux | grep vite
#Or, check the nohup.out file for logs:
cat nohup.out
```

```python
Access the Application:

    By default, Vite’s development server listens on localhost. To access it externally (e.g., from your browser), you need to make it listen on 0.0.0.0:
        Edit your vite.config.js or vite.config.ts in the project root to include:
        javascript

export default {
  server: {
    host: '0.0.0.0',
    port: 5173, // or your preferred port
  },
}
Restart the server after making this change 
```

```python
Ensure your EC2 instance’s Security Group allows inbound traffic on the port Vite is using (e.g., 5173):

    Go to the AWS EC2 Console > Security Groups > Edit Inbound Rules.
    Add a rule for Custom TCP, port 5173, source 0.0.0.0/0 (or restrict to your IP for security).
```

```python
Access the app in your browser at http://<ec2-public-ip>:5173
```

```python
"""
edirect Logs (Optional): If you want to redirect logs to a specific file instead of nohup.out, modify the nohup command:
bash
"""

nohup npm run dev > myapp.log 2>&1 &

"""> myapp.log: Redirects standard output to myapp.log.
2>&1: Redirects standard error to the same file."""
```
